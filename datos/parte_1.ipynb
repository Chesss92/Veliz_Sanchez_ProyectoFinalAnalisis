```python
import os
import sqlite3
import pandas as pd

# 1. Definir la ruta del archivo de base de datos
db_path = r"C:\Users\cris\Desktop\Veliz_Sanchez_ProyectoFinalAnalisis\datos\incidentes.db"

os.makedirs(os.path.dirname(db_path), exist_ok=True)
# Conectar (SQLite creará el archivo automáticamente en esa carpeta)
conn = sqlite3.connect(db_path)
print("¡Conexión establecida con éxito!")

# 2. Configurar la lista de tus archivos
archivos = [
    {
        "mes": "enero",
        "tipo": "csv",
        "path": r"C:\Users\cris\Desktop\Veliz_Sanchez_ProyectoFinalAnalisis\datos\incidentes_enero_2026.csv",
        "sep": ";",
    },
    {
        "mes": "febrero",
        "tipo": "csv",
        "path": r"C:\Users\cris\Desktop\Veliz_Sanchez_ProyectoFinalAnalisis\datos\incidentes_febrero_2026.csv",
        "sep": ";",
    },
    {
        "mes": "marzo",
        "tipo": "csv",
        "path": r"C:\Users\cris\Desktop\Veliz_Sanchez_ProyectoFinalAnalisis\datos\incidentes_marzo_2026.csv",
        "sep": ";",
    },
    {
        "mes": "abril",
        "tipo": "excel",
        "path": r"C:\Users\cris\Desktop\Veliz_Sanchez_ProyectoFinalAnalisis\datos\incidentes_abril_2026.xlsx",
    },
    {
        "mes": "mayo",
        "tipo": "excel",
        "path": r"C:\Users\cris\Desktop\Veliz_Sanchez_ProyectoFinalAnalisis\datos\incidentes_mayo_2026.xlsx",
    },
]

columnas_validas = [
    "Fecha",
    "provincia",
    "Canton",
    "Cod_Parroquia",
    "Parroquia",
    "Servicio",
    "Subtipo",
]

# 3. Insertar datos mes por mes en la tabla 'incidentes'
for item in archivos:
    print(f"Cargando {item['mes']} a SQLite...")

    if item["tipo"] == "csv":
        df = pd.read_csv(item["path"], sep=item["sep"], low_memory=False)
    else:
        df = pd.read_excel(item["path"])

    # Limpiar columnas extra (como las del mes de marzo)
    df = df[[col for col in columnas_validas if col in df.columns]]

    # Formatear la fecha
    df["Fecha"] = pd.to_datetime(df["Fecha"], errors="coerce").dt.strftime(
        "%Y-%m-%d"
    )

    # Insertar en SQLite
    df.to_sql("incidentes", conn, if_exists="append", index=False)

print("¡Consolidación en SQLite realizada con éxito!")
```

```python
import sqlite3
import pandas as pd
from pymongo import MongoClient

# 1. Reabrir la conexión a tu base de datos SQLite
conn = sqlite3.connect(r"C:\Users\cris\Desktop\Veliz_Sanchez_ProyectoFinalAnalisis\datos\incidentes.db")

# 2. Cadena de Conexión a MongoDB (Local)
connection_string = "mongodb://localhost:27017/"
client = MongoClient(connection_string)

# Crear / Seleccionar la Base de Datos y la Colección
db = client["ecu911_db"]
collection = db["incidentes"]

# 3. Leer los datos consolidados desde SQLite
df_completo = pd.read_sql_query("SELECT * FROM incidentes;", conn)

# 4. Convertir el DataFrame a formato de Diccionario (Documentos NoSQL)
documentos = df_completo.to_dict(orient="records")

# 5. Insertar por Lotes (Batch/Chunks)
tamanio_lote = 50000
total_documentos = len(documentos)

print(f"Insertando {total_documentos} documentos en MongoDB...")

# Limpiar la colección si ya tenía datos previos
collection.delete_many({})

for i in range(0, total_documentos, tamanio_lote):
    lote = documentos[i : i + tamanio_lote]
    collection.insert_many(lote)
    print(
        f"Lote insertado: {min(i + tamanio_lote, total_documentos)} / {total_documentos}"
    )

print("\n¡Carga a MongoDB completada exitosamente!")

# Cerrar conexión de SQLite al finalizar
conn.close()
```

```python
import sqlite3
import pandas as pd
from pymongo import MongoClient

# 1. Conectar a SQLite y leer todo el dataset de +2M
conn = sqlite3.connect(r"C:\Users\cris\Desktop\Veliz_Sanchez_ProyectoFinalAnalisis\datos\incidentes.db")
print("Cargando los 2M de registros desde SQLite...")
df_completo = pd.read_sql_query("SELECT * FROM incidentes;", conn)
conn.close()

# 2. Conectar a MongoDB
client = MongoClient("mongodb://localhost:27017/")
db = client["ecu911_db"]
collection = db["incidentes"]

# 3. Limpiar colección previa e insertar en lotes
print("Limpiando colección anterior en MongoDB...")
collection.delete_many({})

print("Insertando documentos en MongoDB por lotes...")
documentos = df_completo.to_dict(orient="records")

tamanio_lote = 100000
total_docs = len(documentos)

for i in range(0, total_docs, tamanio_lote):
    lote = documentos[i : i + tamanio_lote]
    collection.insert_many(lote)
    print(f"Progreso: {min(i + tamanio_lote, total_docs):,} / {total_docs:,}")

print("\n¡Proceso finalizado con éxito en MongoDB!")
```

```python
import os
import random
import sqlite3
import pandas as pd

# 1. Rutas
db_path = "incidentes.db"
csv_ficticios_path = "incidentes_ficticios.csv"

# 2. Conectar a SQLite
conn = sqlite3.connect(db_path)

# Queremos generar exactamente la cuota ficticia necesaria
CANTIDAD_FICTICIOS = 1850000

print(
    f"Generando {CANTIDAD_FICTICIOS:,} registros ficticios obligatoriamente..."
)

# Tomar muestra de la BD actual para mantener nombres de columnas y categorías reales
df_base = pd.read_sql_query("SELECT * FROM incidentes LIMIT 100000;", conn)

# Muestreo aleatorio
df_ficticios = df_base.sample(n=CANTIDAD_FICTICIOS, replace=True).copy()

# Generar variabilidad de fechas sintéticas
fechas_posibles = pd.date_range(
    start="2026-01-01", end="2026-06-30", freq="s"
)
df_ficticios["Fecha"] = [
    random.choice(fechas_posibles).strftime("%Y-%m-%d %H:%M:%S")
    for _ in range(CANTIDAD_FICTICIOS)
]

# 3. GUARDAR EL ARCHIVO CSV
df_ficticios.to_csv(csv_ficticios_path, index=False, encoding="utf-8")

print("\n" + "=" * 50)
print("¡ARCHIVO CREADO CORRECTAMENTE!")
print(f"Nombre del archivo: {csv_ficticios_path}")
print(f"Filas guardadas en CSV: {len(df_ficticios):,}")
print("=" * 50)

conn.close()
```

```python
import os
import random
import sqlite3
import pandas as pd

# 1. Conexión y Meta
db_path = "incidentes.db"
archivo_ficticios = "incidentes_ficticios.csv"
META_TOTAL = 2300000

print("Conectando a la base de datos sqlite...")
conn = sqlite3.connect(db_path)

# 2. Obtener el total de registros limpios en la BD
total_reales = pd.read_sql_query(
    "SELECT COUNT(*) AS total FROM incidentes;", conn
)["total"].iloc[0]
faltantes = META_TOTAL - total_reales

print("\n" + "=" * 50)
print(f"Registros reales en BD:  {total_reales:,}")
print(f"Meta total requerida:   {META_TOTAL:,}")
print(f"Ficticios a generar:    {faltantes:,}")
print("=" * 50 + "\n")

if faltantes > 0:
    print("Cargando muestra base para generar los datos ficticios...")
    df_base = pd.read_sql_query("SELECT * FROM incidentes LIMIT 50000;", conn)

    # Muestreo aleatorio masivo para la diferencia exactas
    df_ficticios = df_base.sample(n=faltantes, replace=True).copy()

    # Aleatorizar fechas para dar variabilidad
    fechas_posibles = pd.date_range(
        start="2026-01-01", end="2026-06-30", freq="s"
    )
    df_ficticios["Fecha"] = [
        random.choice(fechas_posibles).strftime("%Y-%m-%d %H:%M:%S")
        for _ in range(faltantes)
    ]

    # Sobrescribir directamente el CSV
    print(f"Generando y sobrescribiendo '{archivo_ficticios}'...")
    df_ficticios.to_csv(archivo_ficticios, index=False, encoding="utf-8")

    print("\n" + "¡PROCESO COMPLETADO EXITOSAMENTE! " + "✔" * 3)
    print(
        f"El archivo '{archivo_ficticios}' ahora tiene exactamente {len(df_ficticios):,} filas."
    )
else:
    print("Tus datos reales ya alcanzan o superan la meta.")

conn.close()
```
